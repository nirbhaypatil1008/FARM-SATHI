```python
from google.colab import files
from pathlib import Path
import zipfile

export_dir = Path('/content/farm-saathi-model-export')
archive = Path('/content/farm-saathi-model-export.zip')
with zipfile.ZipFile(archive, 'w', compression=zipfile.ZIP_DEFLATED) as bundle:
    for filename in ['plant_disease_model.keras', 'class_labels.json', 'model_metadata.json']:
        bundle.write(export_dir / filename, filename)
files.download(str(archive))
```
